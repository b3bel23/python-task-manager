```python
import rich
from rich.console import Console
from rich.table import Table
from rich.prompt import Prompt, IntPrompt
from rich.panel import Panel
import task_manager

console = Console()

def menu():
    console.rule("[bold magenta]BEM VINDO(A) AO GERENCIADOR DE TAREFAS")
    console.print(Panel.fit("[bold white]1. Criar tarefa", border_style="magenta"))
    console.print(Panel.fit("[bold white]2. Listar tarefas pendentes", border_style="magenta"))
    console.print(Panel.fit("[bold white]3. Marcar tarefa como concluída", border_style="magenta"))
    console.print(Panel.fit("[bold white]4. Remover tarefa", border_style="magenta"))
    console.print(Panel.fit("[bold white]5. Sair", border_style="magenta"))

while True:
    menu()

    opcao = Prompt.ask("Escolha uma opção", choices=["1", "2", "3", "4", "5"])


    if opcao == "1":
        titulo = Prompt.ask("Título da tarefa").strip()
        descricao = Prompt.ask("Descrição (opcional)", default="").strip()
        tarefa = task_manager.create_task(titulo, descricao)
        console.print("[bold green]Tarefa criada com sucesso![/bold green]")

    elif opcao == "2":
        pendentes = task_manager.list_tasks()
        if not pendentes:
            console.print("[yellow]Nenhuma tarefa pendente.[/yellow]")
        else:
            table = Table(title="Tarefas Pendentes")
            table.add_column("Número", justify="right", style="cyan", no_wrap=True)
            table.add_column("Título", style="bold")
            table.add_column("Descrição", style="dim")
            for task in pendentes:
                table.add_row(
                    str(task["number"]),
                    task["title"],
                    task["description"] or "-"
                )
            console.print(table)

    elif opcao == "3":
        try:
            num = IntPrompt.ask("Número da tarefa a marcar como concluída")
            if task_manager.mark_complete(num):
                console.print("[green]Tarefa marcada como concluída.[/green]")
            else:
                console.print("[red]Tarefa não encontrada.[/red]")
        except ValueError:
            console.print("[red]Entrada inválida![/red]")

    elif opcao == "4":
        try:
            num = IntPrompt.ask("Número da tarefa a remover")
            if task_manager.remove_task(num):
                console.print("[green]Tarefa removida com sucesso.[/green]")
            else:
                console.print("[red]Tarefa não encontrada.[/red]")
        except ValueError:
            console.print("[red]Entrada inválida![/red]")

    elif opcao == "5":
        console.print("[bold red]Saindo do programa. Até logo![/bold blue]")
        break
```
